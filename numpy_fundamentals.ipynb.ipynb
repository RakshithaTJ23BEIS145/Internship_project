```python
import numpy as np
import pandas as pd
```

```python
# 1D Array
arr1 = np.array([1, 2, 3, 4, 5])
print("1D Array:")
print(arr1)
print("Shape:", arr1.shape)

# 2D Array
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
print("\n2D Array:")
print(arr2)
print("Shape:", arr2.shape)

# 3D Array
arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("\n3D Array:")
print(arr3)
print("Shape:", arr3.shape)
```

```python
print("Original Array:")
print(arr2)

print("\nAfter Broadcasting (+10):")
print(arr2 + 10)
```

```python
print("Original:", arr1)

print("Multiply by 2:")
print(arr1 * 2)

print("Square:")
print(arr1 ** 2)

print("Square Root:")
print(np.sqrt(arr1))
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = A @ B

print("Matrix A")
print(A)

print("\nMatrix B")
print(B)

print("\nMatrix Multiplication")
print(result)
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

df.head()
```

```python
numeric = df.select_dtypes(include=np.number)

print("Mean")
print(numeric.mean())
```

```python
print("Standard Deviation")
print(numeric.std())
```

```python
print("Correlation Matrix")
print(numeric.corr())
```

```python
print("Mean")
print(numeric.mean())

print("\nStandard Deviation")
print(numeric.std())

print("\nCorrelation")
print(numeric.corr())
```
